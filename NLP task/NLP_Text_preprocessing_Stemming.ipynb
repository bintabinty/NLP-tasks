```python
import pandas as pd
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
```

```python
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
```

```python
df =pd.read_csv("/content/IMDB Dataset.csv",on_bad_lines='skip')
```

```python
df.head()
```

```python
text = df['review']
text.head()
```

```python
text = text.str.lower()

print(text.head())
```

```python
text = text.apply(word_tokenize)

print(text.head())
```

```python
stopwords = set(stopwords.words('english'))
```

```python
cleaned_text = []

for words in text:
    result = []

    for word in words:
        if word not in stopwords:
            result.append(word)

    cleaned_text.append(result)

print(cleaned_text[:5])
```

```python
stemmer = PorterStemmer()

stemmed_text = []

for words in cleaned_text:
    result = []

    for word in words:
        stem = stemmer.stem(word)
        result.append(stem)

    stemmed_text.append(result)

print(stemmed_text[:5])
```

```python
final_text = []

for words in stemmed_text:
    sentence = ' '.join(words)
    final_text.append(sentence)
```

```python
df['stemmed_review'] = final_text
```
