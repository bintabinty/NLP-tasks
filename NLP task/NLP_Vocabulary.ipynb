```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
import csv
```

```python

df=pd.read_csv("/content/Reviews.csv",on_bad_lines='skip', quoting=csv.QUOTE_NONE)
```

```python
df.head()
```

```python
df = df.rename(columns={
    "Text": "Feedback",
    "ProductId": "Product_ID",
    "Score": "Rating"
})
```

```python
df.head()
```

```python
sentences = df["Feedback"]
```

```python
vectorizer = CountVectorizer()
# Fill missing values with empty strings to avoid ValueError
a=vectorizer.fit_transform(sentences.fillna(""))
vocabulary = vectorizer.vocabulary_
```

```python
print(vocabulary)
```

```python
print(a.toarray())
```
